```python
plot_data = wtg150[["wind_speed", "nacelle_position"]].dropna().copy()

# Bin nacelle position into 10-degree intervals
plot_data["nacelle_bin"] = pd.cut(
    plot_data["nacelle_position"],
    bins=range(0, 361, 10)
)

# Average wind speed for each nacelle position bin
bar_data = plot_data.groupby(
    "nacelle_bin",
    observed=True
)["wind_speed"].mean()

# Vertical bar chart
bar_data.plot(kind="bar")

plt.xlabel("Nacelle Position (degrees)")
plt.ylabel("Average Wind Speed (m/s)")
plt.title("Average Wind Speed by Nacelle Position")
plt.xticks(rotation=45)

plt.show()
```
